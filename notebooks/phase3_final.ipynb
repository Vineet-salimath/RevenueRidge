```python
import pandas as pd

# Load only needed columns
cols = [
    "Store",
    "Date",
    "Temperature",
    "Fuel_Price",
    "CPI",
    "Unemployment",
    "IsHoliday",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

df = pd.read_csv(
    "../data/raw/walmart.csv",
    usecols=cols,
    nrows=20000
)

print(df.head())
print(df.shape)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year

print(df[["Date", "Month", "Year"]].head())
```

```python
markdown_cols = [
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

df[markdown_cols] = df[markdown_cols].fillna(0)

print(df.isnull().sum())
```

```python
df.to_csv(
    "../data/cleaned/final_retail_data.csv",
    index=False
)

print("✅ Final dataset saved")
```
